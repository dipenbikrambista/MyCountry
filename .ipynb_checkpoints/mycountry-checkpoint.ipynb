```python
bhutan_df = long_df[long_df['Country Name'] == 'Bhutan'].pivot_table(
    index='Year',
    columns='Series Name',
    values='value'
)

bhutan_df.head()
```

```python
nepal_df = long_df[long_df['Country Name'] == 'Nepal'].pivot_table(
    index= 'Year',
    columns= 'Series Name',
    values= 'value'
)

nepal_df.head()
```

```python
(bhutan_df.corr(numeric_only= True))['Personal remittances, received (% of GDP)'].sort_values(ascending= False)#checking correlation of other metrics with remittance received
# I will choose life expectancy at birth, exports of goods and servies to predict remittance.
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
target = 'Personal remittances, received (% of GDP)'
feature = 'Life expectancy at birth, total (years)'

clean_df_bhutan = bhutan_df.dropna()
X_known = np.array(clean_df_bhutan[[feature]].values)
y_known = np.array(clean_df_bhutan[target].values)

modelremit = LinearRegression()
modelremit.fit(X_known,y_known)
```

```python
r2_score(y_known, modelremit.predict(X_known))
```

```python
#predicting the missing remittance values of Bhutan:
missing_remit = bhutan_df[bhutan_df[target].isna() & bhutan_df[feature].notna()]

X_missing = np.array(missing_remit[[feature]].values)
predicted_remit = modelremit.predict(X_missing)

predicted_years = missing_remit.index

mask = (
    (long_df['Country Name'] == 'Bhutan') &
    (long_df['Series Name'] == target) &
    (long_df['Year'].isin(predicted_years))
)

long_df.loc[mask, 'value'] = predicted_remit
```

## Essay ##

#### Nepal’s major proportion of economic inflow is dependent on remittances. Since many families in Nepal rely on remittances to pay for household expenses, every year, hundreds of thousands of Nepalese leave to work abroad. Although this money has supported many families, and helped improve living standards, it has not translated into comparable growth in Nepal’s domestic economy.

#### The clearest proof of Nepal’s reliance on remittances can be seen in the lineplot comparing its growth with other South Asian countries. As it can be seen from the graph, there has been a rapid increase in remittances (% of GDP) in Nepal. While India, Bangladesh, Pakistan, and Bhutan also experience large amounts of remittances into their economies, none of them has reached the level attained by Nepal (more than 25% of the GDP). This difference is even more striking because Nepal began much closer to its neighbors (roughly around 2% of GDP) before separating sharply around 2001 from them as remittance inflows increased (from roughly 3% to just above 10%). Because of this, remittances isn't merely a way of getting money, but it has become an important aspect of Nepal’s economy. ####

```python
remit_df = long_df[long_df['Series Name'] == 'Personal remittances, received (% of GDP)']
remit_df.head();
```

```python
#plotting the remittance in contrast to other countries for Nepal:
sns.set_style('whitegrid')
plt.figure(figsize=(10,6))
sns.lineplot(data=remit_df, x='Year', y='value', hue='Country Name', linewidth=2)
plt.title("Remittances as % of GDP: Nepal vs. South Asian Neighbors (1993–2024)")
plt.xlim(1993,2024)
plt.xlabel("Year")
plt.ylabel("Remittances (% of GDP)")
plt.legend(title='Country')
plt.show()
```

#### Nepal’s living standards have increased alongside growing reliance on remittances. As it can be seen in the lineplot, the Human Development Index (HDI), a composite index taking into account health, education, and income, has been rising over time. Nepal has also been doing well on other social indicators like life expectancy and a steady decline (comparable to its neighbors) can be seen in the mortality rate of children aged less than five years. 

#### All these improvements are consistent with the benefits provided by the remittances. Remittances by people who work abroad add to the household’s income, which is used for essentials like food, shelter, education, and healthcare services. This, in turn, helps improve the quality of nutrition, healthcare facilities, and education among others.

#### Remittance isn’t an alone factor for these improvements. There are government policies, improvements in healthcare and education systems, and other developments that have helped in improving the situation in Nepal. Nevertheless, the fact that both remittances and HDI have increased along with life expectancy and child mortality rates implies that remittances have been playing a significant role in contributing to the welfare of families. ####

```python
#showing the trend in quality of life in comparision to neighbors:
lifeExp_df = long_df[long_df['Series Name'] == 'Life expectancy at birth, total (years)']

sns.set_style(style= 'whitegrid')
plt.figure(figsize= (10,6))
sns.lineplot(data=lifeExp_df, x = 'Year', y = 'value', hue= 'Country Name') 
plt.title('Life Expectancy at Birth: Nepal vs. South Asian Neighbors (1993–2024)')
plt.xlim(1993,2024)
plt.xlabel('Years')
plt.ylabel('Life expectancy at birth')
plt.legend(title = 'Country')
plt.show()
```

```python
#plotting the Mortality under 5
mortality_df = long_df[long_df['Series Name'] == 'Mortality rate, under-5 (per 1,000 live births)']

plt.figure(figsize=(10,6))
sns.lineplot(data=mortality_df, x='Year', y='value', hue='Country Name', linewidth=2)
plt.title("Under-5 Mortality Rate: Nepal vs. South Asian Neighbors (1993–2024)")
plt.xlabel("Year")
plt.ylabel("Deaths per 1,000 live births")
plt.legend(title='Country')
plt.xlim(1993,2024)
plt.show()
```

```python
#plotting the hdi:
sns.set_style(style= 'whitegrid')
plt.figure(figsize=(10,6))
sns.lineplot(data=hdi_df, x = 'Year', y = 'HDI', hue='Country Name')
plt.title('HDI of Nepal and South Asian Neighbors')
plt.xlabel('Years')
plt.ylabel('Human Development Index')
plt.legend(title = 'Countries')
plt.xlim(1993,2023)
plt.show()
```

#### However, the domestic economy of Nepal has another story to tell. It is clearly depicted through the graphs that both the percentage of exports and manufacturing value added (% of GDP) have dropped dramatically. Manufacturing value (% of GDP) in Nepal has fallen from approximately 9% in the 1990s to 5% in recent times. On the other hand, exports of Nepal have fallen from over 20% of GDP in most parts of the 1990s to below 10% recently. Furthermore, exports and manufacturing value add (% of GDP) both started to decline right after the sudden spike in remittances inflow in the early 2000s. 

#### These trends suggest that Nepal’s increasing remittance income has not translated into comparable growth in its domestic productive sectors. Remittances have the potential of improving living standards and income levels, but earning income abroad does not necessarily result in job creation or increased capacity to export products abroad.The graphs do not prove that remittances caused the decline in manufacturing and exports, but they reveal an important contrast: Nepal is becoming increasingly reliant on foreign income, domestic production is declining.


```python
exports_df = long_df[long_df['Series Name'] == 'Exports of goods and services (% of GDP)']

plt.figure(figsize=(10,6))
sns.lineplot(data=exports_df, x='Year', y='value', hue='Country Name', linewidth=2)
plt.title("Exports as % of GDP: Nepal vs. South Asian Neighbors (1993–2024)")
plt.xlim(1993,2024)
plt.xlabel("Year")
plt.ylabel("Exports (% of GDP)")
plt.legend(title='Country')
plt.show()
```

```python
#Manufacturing value added:
manuf_df = long_df[long_df['Series Name'] == 'Manufacturing, value added (% of GDP)']

plt.figure(figsize=(10,6))
sns.lineplot(data=manuf_df, x='Year', y='value', hue='Country Name', linewidth=2)
plt.title("Manufacturing Value Added as % of GDP: Nepal vs. South Asian Neighbors (1993–2024)")
plt.xlim(1993,2024)
plt.xlabel("Year")
plt.ylabel("Manufacturing (% of GDP)")
plt.legend(title='Country')
plt.show()
```

#### Nepal’s growing dependence on remittances presents both benefits and challenges. The positive impact on the welfare of Nepalese can be seen from the increased life expectancy and HDI and decrease in child mortality rate. However, there are no corresponding improvements in the productivity of the country since the share of industry and exports in GDP is decreasing. Thus, Nepal demonstrates that remittances can enhance the well-being of citizens but cannot replace economic development.

## References ##

[**World Bank, World Developent Indicators**](https://databank.worldbank.org/source/world-development-indicators#) - WDI is compiled by the World Bank from data submitted by national statistical agencies and the IMF's balance-of-payments records, then standardized so that figures are comparable across countries. It is the most widely used reference dataset in academic and policy research on international development, and is cited by economists, governments, and international organizations worldwide.
<br>  
[**Human Development Index Data**](https://hdr.undp.org/data-center/documentation-and-downloads) - The HDI is calculated and published annually by the United Nations Development Programme using a fixed, transparent methodology that combines health, education, and income indicators. It is the standard global measure of human development used in UN reporting.
